##5. What are iterators in Python and how do they differ from iterables?

Answer:

An iterable is an object whose elements can be accessed one by one, such as a list, tuple, or string.

An iterator is an object that returns elements one at a time using next().

Example:

numbers = [10, 20, 30]   # Iterable

it = iter(numbers)       # Iterator

print(next(it))

print(next(it))

Output:

10

20

##6. Explain the concept of generators in Python and how they are defined.

Answer:

A generator is a special type of iterator that produces values one at a time using the yield keyword.

Example:

def numbers():

    yield 1

    yield 2

    yield 3

for n in numbers():

    print(n)

Output:

1

2

3

##7. What are the advantages of using generators over regular functions?

Answer:

The main advantages of generators are:

They use less memory.

They produce values one at a time.

They are useful for large amounts of data.

They are useful for creating sequences.

Example:

def count(n):

    for i in range(1, n + 1):

        yield i

for x in count(3):

    print(x)

##8. What is a lambda function in Python and when is it typically used?

A lambda function is a small anonymous function written in one line using the lambda keyword.

Example:

square = lambda x: x * x

print(square(5))

##9. Explain the purpose and usage of the map() function in Python.

Answer:

The map() function applies a function to every element of an iterable.

Example:

numbers = [1, 2, 3, 4]

result = list(map(lambda x: x * 2, numbers))

print(result)

Output:

[2, 4, 6, 8]

##10. What is the difference between map(), reduce(), and filter() functions in Python?

Answer:

Function	Purpose

map()	Performs an operation on every element

filter()	Selects elements based on a condition

reduce()	Combines all elements into one result

Example:

from functools import reduce

numbers = [1, 2, 3, 4]

print(list(map(lambda x: x * 2, numbers)))

print(list(filter(lambda x: x % 2 == 0, numbers)))

print(reduce(lambda a, b: a + b, numbers))

Output:

[2, 4, 6, 8]

[2, 4]

10

```python
# 1. Write a Python function that takes a list of numbers as input and returns the sum of all even numbers in the list.

def sum_even(numbers):
    total = 0

    for n in numbers:
        if n % 2 == 0:
            total = total + n

    return total

numbers = [1, 2, 3, 4, 5, 6]
print("Sum of even numbers:", sum_even(numbers))
```

```python
# 2. Create a Python function that accepts a string and returns the reverse of that string.

def reverse_string(text):
    return text[::-1]

text = "Python"
print("Reverse:", reverse_string(text))
```

```python
# 3. Implement a Python function that takes a list of integers and returns a new list containing the squares of each number.

def square_numbers(numbers):
    result = []

    for n in numbers:
        result.append(n * n)

    return result

numbers = [1, 2, 3, 4, 5]
print("Squares:", square_numbers(numbers))
```

```python
# 4. Write a Python function that checks if a given number is prime or not from 1 to 200.

def is_prime(n):
    if n < 2 or n > 200:
        return False

    for i in range(2, n):
        if n % i == 0:
            return False

    return True

n = int(input("Enter a number from 1 to 200: "))

if is_prime(n):
    print("Prime number")
else:
    print("Not a prime number")
```

```python
# 5. Create an iterator class in Python that generates the Fibonacci sequence up to a specified number of terms.

def fibonacci(terms):
    a = 0
    b = 1

    for i in range(terms):
        print(a)
        a, b = b, a + b


fibonacci(7)
```

```python
# 6. Write a generator function in Python that yields the powers of 2 up to a given exponent.

def powers_of_two(n):
    for i in range(n + 1):
        yield 2 ** i

for x in powers_of_two(5):
    print(x)
```

```python
# 8. Use a lambda function in Python to sort a list of tuples based on the second element of each tuple.

data = [("A", 3), ("B", 1), ("C", 2)]

result = sorted(data, key=lambda x: x[1])

print(result)
```

```python
# 9. Write a Python program that uses map() to convert a list of temperatures from Celsius to Fahrenheit.

celsius = [0, 10, 20, 30, 40]

fahrenheit = list(map(lambda c: (c * 9 / 5) + 32, celsius))

print("Fahrenheit:", fahrenheit)
```

```python
# 10. Create a Python program that uses filter() to remove all the vowels from a given string.

text = "Hello World"

result = ''.join(filter(lambda x: x.lower() not in "aeiou", text))

print("String without vowels:", result)
```
